```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
trade_history = pd.read_csv('data/processed.csv')
```

```python
x = trade_history.groupby(['Coin','classification']).agg({
    'Closed PnL': ['sum','mean'],
    'Trade ID': 'count'
})
```

```python
x.columns = ['total_pnl', 'avg_pnl', 'trade_count']

x = x.reset_index()
```

## Top Performing Coins Based on Sentiment and PnL

```python

top_profit = x.sort_values(
    'trade_count',
    ascending=False
)

top_5_best = (
    top_profit
    .sort_values(['classification', 'avg_pnl'], ascending=[True, False])
    .groupby('classification')
    .head(5)
)

top_5_best['rank'] = (
    top_5_best.groupby('classification')['avg_pnl']
    .rank(method='first', ascending=False)
)

top_5_best
```
