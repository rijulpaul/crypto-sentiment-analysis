```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
trade_history = pd.read_csv('data/processed.csv')
```

## Week Based Analysis
Day of week based comparison of trade count, pnl mean and sum

```python
# overall totals without classification
overall = trade_history.groupby(
    'day_week'
).agg({
    'Trade ID': 'count',
    'Closed PnL': ['sum', 'mean']
})

overall
```

## Key Insights: Weekly Trading Patterns

- `Wednesday` had the highest trading activity:
  - **36,164 trades**
  - But generated the lowest average pnl:
    - **$16.79 avg pnl**

- `Sunday` was the most profitable trading day:
  - **$63.60 avg pnl**
  - Total pnl:
    - **$1.40M**

- `Tuesday` generated the highest total pnl:
  - **$1.96M**
  - Avg pnl:
    - **$62.11**

- `Thursday` and `Friday` also showed strong profitability:
  - Thursday:
    - **$1.92M total pnl**
  - Friday:
    - **$59.47 avg pnl**

- `Monday` showed relatively weak trading performance:
  - Avg pnl:
    - **$40.82**

- Higher trade activity did not correspond to higher profitability:
  - Example:
    - Wednesday → **36.1K trades**, only **$16.79 avg pnl**
    - Sunday → **22K trades**, but **$63.60 avg pnl**

- Weekend trading (`Saturday–Sunday`) showed stronger average profitability compared to midweek trading periods.
